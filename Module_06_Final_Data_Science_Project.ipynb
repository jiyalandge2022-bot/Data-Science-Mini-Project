```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("All libraries imported successfully!")
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"

df = pd.read_csv(url)

print("Dataset imported successfully!")

display(df.head())
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates().copy()

df = df.dropna()

df["tip_percentage"] = (
    df["tip"] / df["total_bill"]
) * 100

print("Data cleaning completed!")

display(df.head())
```

```python
print("RESTAURANT DATA ANALYSIS")

print("Average Bill:",
      round(df["total_bill"].mean(), 2))

print("Average Tip:",
      round(df["tip"].mean(), 2))

print("Maximum Bill:",
      round(df["total_bill"].max(), 2))

print("Minimum Bill:",
      round(df["total_bill"].min(), 2))

print("Average Tip Percentage:",
      round(df["tip_percentage"].mean(), 2))

print("Average Group Size:",
      round(df["size"].mean(), 2))
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df, x="total_bill",
             bins=20, kde=True)

plt.title("Distribution of Restaurant Bills")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="day", y="tip")

plt.title("Average Tip by Day")
plt.xlabel("Day")
plt.ylabel("Average Tip")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time"
)

plt.title("Relationship Between Bill and Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="time",
    y="tip_percentage"
)

plt.title("Tip Percentage by Dining Time")
plt.xlabel("Dining Time")
plt.ylabel("Tip Percentage")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="day",
    hue="time"
)

plt.title("Restaurant Visits by Day")
plt.xlabel("Day")
plt.ylabel("Number of Visits")

plt.show()
```

```python
print("KEY DATA INSIGHTS")

print("1. Average Restaurant Bill:",
      round(df["total_bill"].mean(), 2))

print("2. Average Tip:",
      round(df["tip"].mean(), 2))

print("3. Highest Average Bill Day:",
      df.groupby("day")["total_bill"].mean().idxmax())

print("4. Highest Average Tip Day:",
      df.groupby("day")["tip"].mean().idxmax())

print("5. Bill and Tip Correlation:",
      round(df["total_bill"].corr(df["tip"]), 3))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

print("Libraries imported successfully!")
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")

display(df.head())
```

```python
print("Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

df = df.drop_duplicates().copy()
df = df.dropna(subset=["tip", "total_bill"]).copy()

print("\nCleaned Dataset:")
display(df.head())
```

```python
print(df.describe())

print("\nAverage Bill:",
      df["total_bill"].mean())

print("\nAverage Tip:",
      df["tip"].mean())

print("\nAverage Group Size:",
      df["size"].mean())
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["total_bill"], kde=True)
plt.title("Restaurant Bill Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="day", y="tip")
plt.title("Average Tips by Day")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="total_bill", y="tip")
plt.title("Relationship Between Bill and Tip")
plt.show()
```

```python
X = df[
    ["total_bill", "size", "sex",
     "smoker", "day", "time"]
]

y = df["tip"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
numeric_features = ["total_bill", "size"]

categorical_features = [
    "sex", "smoker", "day", "time"
]

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

print("Data preprocessing completed!")
```

```python
lr_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])

lr_model.fit(X_train, y_train)

lr_predictions = lr_model.predict(X_test)

print("Linear Regression training completed!")
```

```python
rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=200,
        random_state=42,
        min_samples_leaf=2
    ))
])

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)

print("Random Forest training completed!")
```

```python
def evaluate_model(name, actual, predicted):

    mae = mean_absolute_error(actual, predicted)

    rmse = np.sqrt(
        mean_squared_error(actual, predicted)
    )

    r2 = r2_score(actual, predicted)

    return {
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2 Score": r2
    }


results = []

results.append(
    evaluate_model(
        "Linear Regression",
        y_test,
        lr_predictions
    )
)

results.append(
    evaluate_model(
        "Random Forest",
        y_test,
        rf_predictions
    )
)

results_df = pd.DataFrame(results)

display(results_df.round(3))
```

```python
comparison = pd.DataFrame({
    "Actual Tip": y_test.values,
    "Linear Regression": lr_predictions,
    "Random Forest": rf_predictions
})

display(comparison.head(10))
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    y_test,
    rf_predictions,
    alpha=0.7
)

plt.xlabel("Actual Tips")
plt.ylabel("Predicted Tips")

plt.title("Actual vs Predicted Tips")

plt.show()
```

```python
new_data = pd.DataFrame([{
    "total_bill": 30.00,
    "size": 3,
    "sex": "Female",
    "smoker": "No",
    "day": "Sat",
    "time": "Dinner"
}])

prediction = rf_model.predict(new_data)

print("Predicted Restaurant Tip:",
      round(prediction[0], 2))
```
